## Data Cleaning - Step by Step

We will now address each of the data quality issues identified, ensuring our dataset is robust and reliable for further analysis.

### 1. Handling Missing Values

**Problem:** Missing values (represented as `NaN` or `None`) can lead to biased analyses, incorrect statistical summaries, or errors in machine learning models.

**Recommendation & Suitability:**

*   **Age:** We will impute the missing `Age` values with the **median age**. The median is a robust measure that is less sensitive to outliers compared to the mean, making it a suitable choice for numerical data where the distribution might be skewed.
*   **Fare:** There is only one missing `Fare` value. We will impute this with the **median fare**. Similar to `Age`, the median is a good choice to ensure the imputation is not overly affected by potential extreme values in `Fare`.
*   **Cabin:** Due to a very high percentage of missing values (327 out of 418), simply filling with `'Unknown'` is a practical approach. This treats the absence of a cabin record as a category itself, preventing the loss of information that could occur by dropping the column entirely. It acknowledges that the missingness itself might carry information.

```python
# Impute missing 'Age' values with the median age
# df['Age'].median() calculates the median of the 'Age' column.
# .fillna() replaces NaN values with the calculated median.
# inplace=True modifies the DataFrame directly.
df['Age'].fillna(df['Age'].median(), inplace=True)

# Impute missing 'Fare' values with the median fare
# df['Fare'].median() calculates the median of the 'Fare' column.
# .fillna() replaces NaN values with the calculated median.
# inplace=True modifies the DataFrame directly.
df['Fare'].fillna(df['Fare'].median(), inplace=True)

# Fill missing 'Cabin' values with the string 'Unknown'
# .fillna('Unknown') replaces NaN values in the 'Cabin' column with the string 'Unknown'.
# inplace=True modifies the DataFrame directly.
df['Cabin'].fillna('Unknown', inplace=True)

print("Missing values in 'Age', 'Fare', and 'Cabin' have been handled.")
```

### 2. Handling Duplicate Records

**Problem:** Duplicate rows mean the same observation is recorded multiple times. This can artificially inflate the size of your dataset, lead to skewed statistics, and give disproportionate weight to certain observations in predictive models.

**Recommendation & Suitability:** Removing duplicate rows is generally the most appropriate method. For this dataset, assuming each `PassengerId` should represent a unique person and record, removing exact duplicates ensures data integrity.

**Cleaning Method:** We will use the `drop_duplicates()` method to remove any rows that are exact copies of others. Our initial check showed no duplicates, but it's good practice to include this step.

```python
# Calculate the number of duplicate rows in the DataFrame
# df.duplicated() returns a boolean Series where True indicates a row is a duplicate of a previous row.
# .sum() then counts these True values.
duplicate_rows_count = df.duplicated().sum()

print(f"Number of duplicate rows found: {duplicate_rows_count}")

# If duplicate rows are found, remove them
if duplicate_rows_count > 0:
    # df.drop_duplicates() creates a new DataFrame with duplicate rows removed.
    # By default, it keeps the first occurrence and removes subsequent ones.
    # We assign the result back to df to update the DataFrame.
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows to remove.")

# Verify that duplicate rows have been removed by checking the count again
# This should ideally be 0 after removal.
duplicate_rows_count_after = df.duplicated().sum()
print(f"Number of duplicate rows after removal: {duplicate_rows_count_after}")

# Display the new shape of the DataFrame to show the reduction in rows if duplicates were present
print(f"\nNew shape of the DataFrame after handling duplicates: {df.shape}")
```

### 3. Correcting Data Types

**Problem:** Data types that don't accurately reflect the nature of the data can lead to inefficient memory usage, incorrect operations (e.g., mathematical operations on categorical data), or hinder certain analyses.

**Recommendation & Suitability:**

*   **Age (float to int):** After imputing missing `Age` values, the column is currently `float64`. Since age is typically a whole number, converting it to an integer type (`int`) is more appropriate and memory-efficient.
*   **Sex (object to category):** `Sex` (male/female) is a categorical variable with a limited number of unique values. Converting it to the `category` data type can significantly reduce memory usage and optimize operations for categorical data.
*   **Embarked (object to category):** Similar to `Sex`, `Embarked` (port of embarkation) is also categorical. Converting it to `category` type offers the same benefits.

**Cleaning Method:** We will use the `.astype()` method to explicitly change the data types.

```python
# Convert the 'Age' column to integer type
# The .astype(int) method is used to change the data type of the column to integer.
# This makes sense as age is usually represented as a whole number.
df['Age'] = df['Age'].astype(int)

# Convert 'Sex' and 'Embarked' columns to 'category' data type
# Converting object (string) types to 'category' can save memory and is appropriate for fixed, limited sets of values.
# This also helps pandas recognize these columns as true categorical variables for analysis.
df['Sex'] = df['Sex'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')

# Convert 'Cabin' to 'category' data type after filling missing values
# This is appropriate since 'Unknown' acts as a new category.
df['Cabin'] = df['Cabin'].astype('category')

print("Data types refined for 'Age', 'Sex', 'Embarked', and 'Cabin'.")

# Verify the changes by displaying the updated data types for all columns
print("\nUpdated Data Types:")
display(df.dtypes)

# Display the first few rows of the DataFrame to observe the changes in 'Age' type (e.g., no decimal places)
print("\nDataFrame head after refining data types:")
display(df.head())
```

### 4. Checking for Inconsistent Values (Categorical Data)

**Problem:** Inconsistent spellings, different casing (e.g., 'male' vs 'Male'), or variations of the same category can lead to miscounted categories and incorrect analysis. For example, 'S' and 'Southampton' for the same port.

**Recommendation & Suitability:** Inspecting the unique values and their counts (`value_counts()`) for categorical columns helps identify inconsistencies. If found, standardizing them (e.g., converting to lowercase, mapping variations to a single correct value) is crucial. For this dataset, 'Sex', 'Embarked', and 'Cabin' are the primary candidates for this check.

**Cleaning Method:** We will examine the `value_counts()` for 'Sex', 'Embarked', and 'Cabin' to ensure consistency.

```python
# Check for inconsistencies in the 'Sex' column by looking at its unique values and their counts
# .value_counts() provides a Series containing counts of unique values in descending order.
print("Unique values and counts for 'Sex':")
display(df['Sex'].value_counts())

# Check for inconsistencies in the 'Embarked' column by looking at its unique values and their counts
print("\nUnique values and counts for 'Embarked':")
display(df['Embarked'].value_counts())

# Check for inconsistencies in the 'Cabin' column by looking at its unique values and their counts
print("\nUnique values and counts for 'Cabin':")
display(df['Cabin'].value_counts())

print("\nBased on the output, 'Sex' and 'Embarked' appear to be consistent. 'Cabin' shows many unique values, as expected, with 'Unknown' being the most frequent due to imputation.")
```

### 5. Identifying Potential Outliers

**Problem:** Outliers are data points that significantly differ from other observations. They can be genuine extreme values or errors. Outliers can heavily influence statistical measures (like mean and standard deviation) and negatively impact the performance of some machine learning models.

**Recommendation & Suitability:** Identifying outliers often involves visualization (like box plots or histograms) or statistical methods (e.g., Z-scores, IQR method). For `Fare` and `Age`, a visual inspection using box plots is a good starting point to see their distribution and identify any extreme values. For now, we will identify them and decide on treatment later, as outlier treatment depends on the analysis goal and the specific modeling approach.

**Cleaning Method:** We will use box plots to visualize the distributions of 'Age' and 'Fare' to spot potential outliers.

```python
# Create a figure with two subplots side-by-side
plt.figure(figsize=(15, 6))

# Subplot 1: Box plot for the 'Age' column to visualize its distribution and potential outliers
# A box plot shows the median, quartiles, and potential outliers (points beyond the whiskers).
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.boxplot(y=df['Age'])
plt.title('Box Plot of Age')
plt.ylabel('Age')

# Subplot 2: Box plot for the 'Fare' column to visualize its distribution and potential outliers
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.boxplot(y=df['Fare'])
plt.title('Box Plot of Fare')
plt.ylabel('Fare')

plt.tight_layout() # Adjust layout to prevent overlapping titles/labels
plt.show()

print("Box plots for 'Age' and 'Fare' displayed to identify potential outliers. Further decisions on handling outliers would depend on the specific analysis goals.")
```

#### Visualization 2.1: Correlation Heatmap of Numerical Variables

*   **Reason for Choice**: A correlation heatmap provides a concise visual summary of the linear relationships between all pairs of numerical variables, including the target `Survived`. It helps quickly identify which numerical features are strongly correlated with each other and with survival, serving as a foundational step in multivariate analysis.

```python
# Select only relevant numerical columns for the correlation matrix.
# 'Survived' is included as it's our numerical target variable.
# 'PassengerId' is excluded as it is a unique identifier, not a measurable feature.
numerical_features = ['Survived', 'Age', 'Fare', 'SibSp', 'Parch', 'Pclass']

# Calculate the pairwise correlation between selected numerical features.
# The .corr() method computes the standard correlation coefficient.
correlation_matrix = df[numerical_features].corr()

# Create a heatmap to visualize the correlation matrix.
plt.figure(figsize=(10, 8)) # Set the size of the plot for optimal readability.
sns.heatmap(correlation_matrix,
            annot=True,     # Display the correlation coefficients on the heatmap cells.
            cmap='coolwarm',  # Use a diverging colormap ('coolwarm' shows positive/negative correlations).
            fmt='.2f',      # Format the annotations to two decimal places.
            linewidths=.5)  # Add lines between cells for better visual separation.
plt.title('Correlation Heatmap of Numerical Features', fontsize=16) # Set a descriptive chart title.
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better legibility.
plt.yticks(rotation=0) # Ensure y-axis labels are horizontal.
plt.show() # Display the generated plot.

print("\n--- Interpretation ---")
print("The heatmap visually represents the strength and direction of linear relationships between numerical variables. Values range from -1 (strong negative correlation) to +1 (strong positive correlation), with 0 indicating no linear correlation.")
print("\n--- Key Relationships Revealed ---")
print("-   **'Fare' and 'Survived' (0.19)**: There is a moderate positive correlation, suggesting that passengers who paid higher fares had a somewhat better chance of survival.")
print("-   **'Pclass' and 'Survived' (-0.11)**: A weak negative correlation exists, indicating that lower passenger class (e.g., 3rd class, which has a higher numerical value) is associated with a slightly lower chance of survival. This aligns with the 'higher class, higher survival' expectation.")
print("-   **'Age' and 'Survived' (0.01)**: A very weak positive correlation, suggesting age alone has little linear relationship with survival. This might change when considering interactions with other variables.")
print("-   **'Fare' and 'Pclass' (-0.58)**: A strong negative correlation, indicating that higher fares are associated with lower passenger classes (e.g., 1st class, which has a lower numerical value). This is an expected relationship as higher classes generally cost more.")
print("-   **'Age' and 'Pclass' (-0.47)**: A moderate negative correlation, implying that passengers in higher classes tend to be older on average, or conversely, lower class passengers are generally younger.")
print("Overall, the heatmap confirms expected relationships and highlights that 'Fare' and 'Pclass' have the most substantial linear associations with 'Survived' among the numerical features.")
```

#### Visualization 2.2: Survival Rate by `Pclass` and `Sex`

*   **Reason for Choice**: This visualization was chosen to investigate the combined effect of two of the strongest predictors identified in bivariate analysis: `Pclass` and `Sex`. It allows us to observe if the survival advantage for females was consistent across all passenger classes, or if social class amplified or diminished this effect. This chart helps to reveal powerful interactions that simple individual analyses cannot.

```python
# Calculate the mean survival rate grouped by 'Pclass' and 'Sex'.
# This aggregates the data to show the proportion of survivors (mean of 'Survived')
# for each unique combination of passenger class and gender.
# setting observed=False ensures that all categories, even those not present in a subgroup, are shown.
mean_survival_pclass_sex = df.groupby(['Pclass', 'Sex'], observed=False)['Survived'].mean().reset_index()

# Create a grouped bar plot to visualize the survival rate.
# 'x' represents Pclass on the horizontal axis.
# 'y' represents the calculated mean 'Survived' (survival rate) on the vertical axis.
# 'hue' is set to 'Sex' to create separate bars for 'male' and 'female' within each Pclass group.
plt.figure(figsize=(10, 7)) # Set the size of the plot for better readability.
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=mean_survival_pclass_sex, palette='viridis')
plt.title('Survival Rate by Passenger Class and Sex', fontsize=16) # Set a descriptive chart title.
plt.xlabel('Passenger Class') # Set the x-axis label.
plt.ylabel('Survival Rate (Proportion)') # Set the y-axis label.
# Customize x-axis tick labels for clarity.
plt.xticks(ticks=[0, 1, 2], labels=['1st Class', '2nd Class', '3rd Class'])
plt.legend(title='Sex') # Add a legend to distinguish between male and female bars.
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a horizontal grid for easier comparison of rates.
plt.show() # Display the generated plot.

print("\n--- Interpretation ---")
print("This grouped bar chart clearly illustrates the interplay between passenger class and gender on survival. It vividly demonstrates the 'women and children first' policy in action.")
print("\n--- Key Relationships Revealed ---")
print("-   **Female Advantage**: Females consistently had a significantly higher survival rate than males across all passenger classes. This confirms the strong effect of 'Sex' on survival.")
print("-   **Class Impact**: Within both genders, passengers in higher classes generally had better survival rates. For example, 1st class females had nearly 100% survival, while 3rd class females, although still much higher than males, had a lower survival rate than their higher-class counterparts.")
print("-   **Striking Interaction**: The most striking interaction is that 3rd class females still had a higher survival rate than 1st class males. This highlights that being female was a stronger predictor of survival than being in a higher social class, although both were significant factors.")
print("-   **Disparity for Males**: Males, particularly in 3rd class, faced extremely low survival rates, reinforcing the selective nature of evacuation.")
```

#### Visualization 2.3: `Age` Distribution by `Pclass`, `Sex`, and `Survived`

*   **Reason for Choice**: This visualization allows for a detailed examination of how `Age` distribution varies across different combinations of `Pclass` and `Sex`, with `Survived` as a further distinguishing factor. It is crucial for understanding nuanced survival patterns, such as the 'children first' policy, in conjunction with the 'women first' and 'class privilege' observations. A `catplot` with `kind='box'` is ideal for showing the statistical distribution of a numerical variable across multiple categorical groupings.

```python
# Create a grouped box plot to visualize Age distribution by Pclass, Sex, and Survived.
# sns.catplot is used here because it effectively handles faceting (creating separate subplots for 'col' variable).
# 'x' represents 'Pclass', 'y' represents 'Age', 'hue' differentiates by 'Survived' status,
# and 'col' creates separate subplots for each 'Sex'.
# kind='box' specifies that the plot type within each facet should be a box plot.

g = sns.catplot(x='Pclass', y='Age', hue='Survived', col='Sex', data=df, kind='box',
                height=6, aspect=1.2, palette='coolwarm') # Adjust height and aspect for better readability and layout.
g.set_axis_labels('Passenger Class', 'Age') # Set common axis labels for the entire plot.
g.set_titles('Sex = {col_name}') # Set titles for each facet (column) to indicate the gender.

# Customize x-axis labels for clarity on each subplot.
new_xticks = ['1st Class', '2nd Class', '3rd Class']
for ax in g.axes.flat:
    ax.set_xticks(range(len(new_xticks))) # Set numerical tick locations explicitly.
    ax.set_xticklabels(new_xticks)

g._legend.set_title('Survived') # Set the title for the hue legend.
# Update legend labels to be more descriptive.
for t, l in zip(g._legend.texts, ['Did Not Survive', 'Survived']):
    t.set_text(l)

plt.suptitle('Age Distribution by Passenger Class, Sex, and Survival Status', y=1.03, fontsize=16) # Set an overall title for the entire figure.
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent titles and labels from overlapping.
plt.show() # Display the generated plot.

print("\n--- Interpretation ---")
print("This grouped box plot provides a detailed view of how age, class, sex, and survival interact. Each box represents the age distribution for a specific group (e.g., 1st class females who survived).")
print("\n--- Key Relationships Revealed ---")
print("-   **'Children First' Visible**: Across all classes and for both sexes, younger children (indicated by the lower age values in the box plots) show higher survival rates. This is especially evident in 2nd and 3rd class, where the median age of survivors is notably lower than non-survivors among males.")
print("-   **Female Survival Across Age**: For females, particularly in 1st and 2nd class, survival rates are high across most age groups, implying that gender was a predominant factor overriding age in these categories.")
print("-   **Male Disparity**: Male survival is generally lower, but the 'children first' effect is more pronounced. Young boys in higher classes had a better chance than adult males.")
print("-   **Class and Age Interaction**: In 3rd class, both male and female children (younger ages) had a better chance of survival compared to adults within the same class and gender, although overall 3rd class survival remained lower.")
```

### Executive Summary: Analysis of Titanic Passenger Survival Factors

**Purpose:** This analysis was conducted to identify and understand the key demographic and socio-economic factors that influenced passenger survival during the Titanic disaster. The objective was to derive actionable insights from historical data that could inform modern safety protocols and emergency preparedness.

**Dataset Description:** The study utilized a dataset comprising detailed information for 418 Titanic passengers, including their survival status, gender, passenger class, age, fare paid, number of siblings/spouses aboard, number of parents/children aboard, and port of embarkation. The data underwent rigorous cleaning to address missing values and ensure data integrity.

**Most Important Findings:**

The analysis revealed several critical factors impacting survival:

1.  **Gender as the Primary Determinant:** Female passengers consistently demonstrated significantly higher survival rates across all passenger classes. This pattern was so pronounced that even third-class females had a better chance of survival than first-class males, underscoring a clear prioritization of women during the evacuation.
2.  **Socio-Economic Influence:** Passenger class and fare paid were strongly correlated with survival. Individuals in higher classes (1st and 2nd) exhibited markedly greater survival probabilities, indicating that socio-economic status played a substantial role in access to safety resources.
3.  **Age-Related Prioritization:** Younger children, irrespective of their gender or class, showed higher survival rates, highlighting a 'children first' policy during the rescue efforts. This was particularly evident among males in lower classes.
4.  **Family Unit Impact:** Passengers traveling in small family units (e.g., 1-2 siblings/spouses or 1-3 parents/children) generally had better survival rates compared to those traveling alone or in very large groups, suggesting an optimal family size for support during emergencies.

**Overall Business Implications:** The findings underscore the critical importance of well-defined and rigorously enforced emergency protocols. The stark disparities in survival rates based on gender and socio-economic status reveal how ingrained biases and unequal access to safety resources can manifest in tragic outcomes. These insights are invaluable for developing more equitable and effective safety training, resource allocation strategies, and operational plans for emergency situations in any context where human lives are at stake. Understanding these historical patterns can guide the creation of inclusive and robust safety frameworks.
